### Problem 1

**Python code to categorize an integer using nested if statements**

-    Take user input in the form of integer
-    Find out and print if it is even or odd also if it is positive or negative or zero

```python
# Solution Problem 1:
num = int(input("Enter an integer: ")) # Take input from user
if num == 0: # Check if number is zero first
    print("The number is Zero.")
else:
    if num > 0: # Check positive or negative
        print("The number is Positive.")
    else:
        print("The number is Negative.")

    # Now check even or odd
    if num % 2 == 0:
        print("It is Even.")
    else:
        print("It is Odd.")
```

### Problem 2

**Python code to print the length of a string**

-    Ask name from user in the form of string
-    Print the name of user and along with the length of provided string(name)

```python
# Solution Problem 2: Print the length of a string

# Take name input from user
name = input("Enter your name: ")

# Find length using len()
length = len(name)

# Print name and its length
print("Your name is:", name)
print("The length of your name is:", length)
```

### Problem 3
Print the alpahbet of string

*   Assign the string "Hello world" to a variable named "x"
*   Update the variable name to "y" for the same string
*   Print the alphabet present at -1 index
*   Print the alphabet present at 0 index





```python
# Problem 3: Print the alphabet of string

# Assign string to variable x
x = "Hello world"

# Update variable name to y
y = x

# Print alphabet at -1 index (last character)
print("Alphabet at -1 index:", y[-1])

# Print alphabet at 0 index (first character)
print("Alphabet at 0 index:", y[0])
```

### Problem 5

**Check cookie order conditions**

Assign the number 8 to a variable named **order** (this means a customer ordered 8 cookies)

Assign the number 12 to a variable named **stock** (this means the bakery has 12 cookies in stock)

Write an if statement that checks two conditions together using and:

*    The order is 12 or fewer cookies

*    The stock is greater than or equal to the order

If both conditions are *true*, print "Order can be fulfilled"

Otherwise, print "Order cannot be fulfilled"

```python
# Solution Problem 5: Check cookie order conditions

# Assign values
order = 8    # customer ordered 8 cookies
stock = 12   # bakery has 12 cookies

# Check conditions using 'and'
if order <= 12 and stock >= order:
    print("Order can be fulfilled")
else:
    print("Order cannot be fulfilled")
```

### Problem 6

**Check exam result**

*    Ask the user to enter their exam score (score)

*    If the score is 50 or more, print "You passed!"
